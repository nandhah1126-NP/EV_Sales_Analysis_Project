```python
import matplotlib.pyplot as plt
brand_sales = df.groupby("brand")["annual_sales_units"].sum().sort_values(ascending=False)
brand_sales.plot(kind="bar",figsize=(12,6))
plt.title("Total ev sales by brand")
plt.xlabel("Brand")
plt.ylabel("Annual_sales_units")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

brand_sales = (df.groupby("brand")["annual_sales_units"].sum().sort_values(ascending=False))
brand_sales.plot(kind="bar",figsize=(12, 6))
plt.title("total ev sales by brands")
plt.xlabel("brand")
plt.ylabel("annual sales (Milloins)")
plt.xticks(rotation=60)
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x , pos : f'{x/1e6:.0f}M'))
plt.show()
```

```python
Body_Sales=(df.groupby("body_type")["annual_sales_units"].sum().sort_values(ascending=False))
Body_Sales.plot(kind="bar",figsize=(6,4))

plt.title("Total Ev Bales by Body Types")
plt.xlabel("Body_Types")
plt.ylabel("Annual_Sales_Units")
plt.xticks(rotation=0)

plt.show()
```

```python
Country_Sales= (df.groupby("country_of_origin")["annual_sales_units"].sum().sort_values(ascending=False))
Country_Sales.plot(kind="bar",figsize=(8,5))

plt.title("Total Ev Sales by Country")
plt.xlabel("Country_of_Origin")
plt.ylabel("Annual_Sales_Units")
plt.xticks(rotation=45)

plt.show()
```

```python
Segment_Sales=(df.groupby("market_segment")["annual_sales_units"].sum().sort_values(ascending=False))
Segment_Sales.plot(kind="bar",figsize=(8,5))
plt.title("Annual Ev Sales by Market Segment")
plt.xlabel("Market_Segment")
plt.ylabel("Annual_Sales_Units")
plt.xticks(rotation=0)
plt.show()
```

```python
Year_Sales=(df.groupby("year")["annual_sales_units"].sum())
Year_Sales.plot(kind="line",marker="o",figsize=(10,5))
plt.title("Annual Ev Sales by Year")
plt.xlabel("Year")
plt.ylabel("Annual_Sales_Units")
plt.xticks(rotation=0)
plt.show()
```

```python
avg_price=df.groupby("brand")["price_usd"].mean().sort_values(ascending=False)
avg_price.plot(kind="bar",figsize=(8,5))
plt.title("Average Ev Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Price_Usd")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["battery_capacity_kwh"],bins=15)
plt.title("Battery Capacity Distribution")
plt.xlabel("Battery_Capacity")
plt.ylabel("Number of Vehicles")
plt.show()
```

```python
correlation=df.corr(numeric_only=True)
plt.figure(figsize=(12,8))
plt.imshow(correlation,cmap="coolwarm")
plt.colorbar()
plt.xticks(range(len(correlation.columns)),correlation.columns,rotation=90)
plt.yticks(range(len(correlation.columns)),correlation.columns)
plt.title("Correlation Heatmap")
plt.show()
```
